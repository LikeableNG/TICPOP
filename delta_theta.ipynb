```python
import pandas as pd
import io

caminho_arquivo = '/home/likeableng/BeACoN/arquivos_de_reducao/18abr11/reduced/hd148937/Obs_2_hd148937/resultados_polarimetria_obs2_hd148937.out'

dados_filtrados = []

# 1. Lendo o arquivo com o encoding correto e limpando o cabeçalho
with open(caminho_arquivo, 'r', encoding='utf-8') as f:
    for linha in f:
        linha_limpa = linha.strip()
        
        # Correção da sintaxe do IF simplificado
        if linha_limpa.startswith('B ') or linha_limpa.startswith('V '):
            # Remove o caractere '±' para que a divisão por espaços funcione perfeitamente
            linha_formatada = linha_limpa.replace('±', '')
            dados_filtrados.append(linha_formatada)

# Une as linhas filtradas em um único bloco de texto
conteudo_tabela = "\n".join(dados_filtrados)

# 2. Definindo os nomes das colunas separadas
colunas = [
    'Filtro', 'N_Usados', 
    'P_(%)', 'P_Erro', 
    'Theta_(°)', 'Theta_Erro', 
    'k', 'RMS'
]

# 3. Criando o DataFrame usando 'sep' atualizado para evitar o aviso de depreciação
df = pd.read_csv(io.StringIO(conteudo_tabela), sep=r'\s+', names=colunas)

# 4. Salvando em CSV no mesmo diretório ou localmente
df.to_csv('resultado_polarimetria.csv', index=False)

print("Conversão concluída com sucesso! Arquivo 'resultado_polarimetria.csv' gerado.")

# Define o caminho completo para salvar na mesma pasta do arquivo original
caminho_salvamento = '/home/likeableng/BeACoN/arquivos_de_reducao/18abr11/reduced/hd148937/Obs_2_hd148937/resultado_polarimetria.csv'

df.to_csv(caminho_salvamento, index=False)
print(f"Arquivo salvo com sucesso em: {caminho_salvamento}")

```
